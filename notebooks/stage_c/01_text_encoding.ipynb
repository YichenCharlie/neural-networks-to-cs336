```python
text = "hello world"

# 1. 去重并排序，建立词表
unique_chars = set()
for char in text:
    unique_chars.add(char)

chars = sorted(unique_chars)
vocab_size = len(chars)

print("词表:", chars)
print("词表大小:", vocab_size)

# 2. 建立双向映射
stoi = {}
itos = {}

for i, char in enumerate(chars):
    stoi[char] = i
    itos[i] = char

print("字符 → ID:", stoi)
print("ID → 字符:", itos)

# 3. 编码：文本 → ID 列表
encoded = []

for char in text:
    encoded.append(stoi[char])

print("编码结果:", encoded)

# 4. 解码：ID 列表 → 文本
decoded_chars = []

for token_id in encoded:
    decoded_chars.append(itos[token_id])

decoded = "".join(decoded_chars)

print("还原文本:", decoded)
print("是否与原文一致:", decoded == text)
```

```python
import torch

x = torch.tensor(encoded[:-1], dtype= torch.long)
y = torch.tensor(encoded[1:], dtype= torch.long)

print(x)
print(y)
print(x.shape)
print(y.shape)

for i in x:
    print(itos[int(i)],end="")
print("\n")
for i in y:
    print(itos[int(i)],end="")


```

## Bigram：通过 Embedding 查表得到预测分数

```python
import torch.nn as nn

table = nn.Embedding(num_embeddings=8,embedding_dim=8)
pred = table(x)
print(table.weight.shape,x.shape,pred.shape)
print(torch.allclose(pred[0],table.weight[x[0]]))
print(pred.argmax(dim=1))
```

```python
loss_api = nn.CrossEntropyLoss()
loss = loss_api(pred,y)
print("loss:",loss)
print("loss.shape",loss.shape)
print(loss.requires_grad)
```
