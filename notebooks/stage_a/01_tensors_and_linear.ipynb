```python
import torch
X = torch.tensor([[1,2],[3,4],[-1,-2],[0,0]])
W = torch.tensor([[1,0,-1],[2,1,0]])
```

```python
b = torch.tensor([0.5,-0.5,1])
```

```python
Z = X @ W
```

```python
Y = torch.add(Z,b)
```

```python
print(Y)
```

```python
print(X.shape)
print(W.shape)
print(b.shape)
print(X.dtype)
print(W.dtype)
print(b.dtype)
```
