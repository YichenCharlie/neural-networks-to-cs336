```python
import torch
import matplotlib.pyplot as plt
```

```python
y = torch.tensor([1,3,5,7], dtype = torch.float32)
x = torch.tensor([0,1,2,3], dtype = torch.float32)
w = torch.tensor(0, dtype = torch.float32, requires_grad = True)
b = torch.tensor(0, dtype = torch.float32, requires_grad = True)
```

```python
y_predict = x * w + b
squared_errors = (y - y_predict) ** 2
loss = squared_errors.mean()
```

```python
n = 0.01
loss_arr = []
with torch.no_grad():
    initial_loss = ((x * w + b - y) ** 2).mean()
    loss_arr.append(initial_loss.item())

for step in range(1000):
    w.grad = None
    b.grad = None
    y_predict = x * w + b
    squared_errors = (y - y_predict) ** 2
    loss = squared_errors.mean()
    loss.backward()
    with torch.no_grad():
        w -= w.grad * n 
        b -= b.grad * n
        y_predict = x * w + b
        squared_errors = (y - y_predict) ** 2
        loss = squared_errors.mean()
        loss_arr.append(loss.item())

        if (step + 1) % 100 == 0:
            print(
            f"Step {step + 1}: "
            f"w={w.item():.4f}, b={b.item():.4f}, "
            f"loss={loss.item():.6f}"
            )
            print("Predictions:", y_predict.tolist())
```

```python
print("Initial loss:", loss_arr[0])
print("Recorded states:", len(loss_arr))
```

训练时根据 loss 求梯度并更新参数；推理时使用已经学到的参数计算输出。
