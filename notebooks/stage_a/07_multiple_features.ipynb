```python
import torch
import torch.nn as nn
```

```python
y = torch.tensor([[1],[3],[-2],[0],[5],[-5]],dtype= torch.float32)
x = torch.tensor([[0,0],[1,0],[0,1],[1,1],[2,0],[0,2]],dtype= torch.float32)
w1 = torch.tensor(0,dtype= torch.float32,requires_grad= True)
w2 = torch.tensor(0,dtype= torch.float32,requires_grad= True)
b = torch.tensor(0,dtype= torch.float32,requires_grad= True)
layer = nn.Linear(2,1)

optimzer = torch.optim.SGD(layer.parameters(),lr = 0.01)

shousuan = x @ layer.weight.T + layer.bias

print(torch.allclose(shousuan,layer(x)))
print(x.shape)
print(y.shape)
print(layer.bias.shape)
print(layer.weight.shape)

optimzer.zero_grad(set_to_none= True)
y_pred = layer(x)
loss = ((y - y_pred) ** 2).mean()
print("Prediction shape:", y_pred.shape)
print("Loss shape:", loss.shape)
print("Loss:", loss.item())
# loss.backward()
# optimzer.step()
```

```python
for step in range(1000):
    optimzer.zero_grad(set_to_none=True)
    y_predict = layer(x)
    loss = ((y - y_predict) ** 2).mean()
    loss.backward()
    optimzer.step()
    with torch.no_grad():
        y_predict = layer(x)
        loss_after = ((y - y_predict) ** 2).mean()

    if (step + 1) % 200 == 0:
        print(
                f"step {step + 1}:"
                f"w={layer.weight.tolist()}, b = {layer.bias.item():.4f},"
                f"loss = {loss_after.item():.6f}"
            )
        print("Prediction:",y_predict.tolist())
```
