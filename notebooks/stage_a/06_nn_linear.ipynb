```python
import torch
import matplotlib.pyplot as plt
import torch.nn as nn
```

```python
x = torch.tensor([[0],[1],[2],[3]],dtype = torch.float32)
y = torch.tensor([[1],[3],[5],[7]],dtype = torch.float32)
w = torch.tensor(0, dtype = torch.float32, requires_grad = True)
b = torch.tensor(0, dtype = torch.float32, requires_grad = True)
```

```python
layer = nn.Linear(in_features = 1 , out_features = 1)

optimizer = torch.optim.SGD(layer.parameters(),lr = 0.01)

for step in range(1000):
    optimizer.zero_grad(set_to_none = True)
    y_predict = layer(x)
    loss = ((y - y_predict) ** 2).mean()
    loss.backward()
    optimizer.step()

    
    if (step + 1) % 100 == 0:
        with torch.no_grad():
            y_predict = layer(x)
            loss_after = ((y - y_predict) ** 2).mean()
        print(
            f"step {step + 1}:"
            f"w={layer.weight.item():.4f}, b = {layer.bias.item():.4f},"
            f"loss = {loss_after.item():.6f}"
        )
        print("Prediction:",y_predict.tolist())

    
```
