```python
import torch
import matplotlib.pyplot as plt
```

```python
y = torch.tensor([1,3,5,7], dtype = torch.float32)
x = torch.tensor([0,1,2,3], dtype = torch.float32)
w = torch.tensor(0, dtype = torch.float32, requires_grad = True)
b = torch.tensor(0, dtype = torch.float32, requires_grad = True)
```

```python
y_predict = x * w + b
squared_errors = (y - y_predict) ** 2
loss = squared_errors.mean()
```

```python
optimizer = torch.optim.SGD([w],lr = 0.01)
optimizer.zero_grad(set_to_none=True)
loss.backward()
optimizer.step()
y_predict = x * w + b
squared_errors = (y - y_predict) ** 2
loss = squared_errors.mean()
print(w.item(),b.item())
print("predict:",y_predict)
```

```python
optimizer = torch.optim.SGD([w,b] , lr = 0.01)
for step in range(1000):
    optimizer.zero_grad(set_to_none=True)
    loss.backward()
    optimizer.step()
    y_predict = w * x + b
    squared_errors = (y - y_predict) ** 2
    loss = squared_errors.mean()
    if (step + 1) % 100 == 0:
        print(
            f"Step {step + 1}: "
            f"w={w.item():.4f}, b={b.item():.4f}, "
            f"loss={loss.item():.6f}"
            )
        print("Predictions:", y_predict.tolist())
```
