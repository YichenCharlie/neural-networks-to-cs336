```python
import torch
import torch.nn as nn
from torchvision import datasets,transforms
from torch.utils.data import DataLoader
```

```python
train_dataset = datasets.FashionMNIST(
    root="../../data",
    train=True,
    download=True,
    transform=transforms.ToTensor(),
)

loaded_data = DataLoader(train_dataset,batch_size=32,shuffle= True,num_workers= 0 )
images , labels= next(iter(loaded_data))
flat = images.flatten(start_dim = 1)
print(flat.shape)
```

```python
class imageMLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(784,128)
        self.layer2 = nn.Linear(128,10)

    def forward(self, x):
        x = x.flatten(start_dim = 1)
        y1 = self.layer1(x)
        y2 = torch.relu(y1)
        y3 = self.layer2(y2)
        return y3
        
```

```python
model = imageMLP()
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
loss_api = nn.CrossEntropyLoss()

epochs = 1
model.train()

for epoch in range(epochs):
    correct_num = 0
    total_num = 0
    total_loss = 0.0
    for batch_index, (images, labels) in enumerate(loaded_data):
        optimizer.zero_grad(set_to_none=True)

        prediction = model(images)

        
        the_label = prediction.argmax(dim= 1)
        correct_num += (the_label == labels).sum().item()
        total_num += labels.shape[0]

        loss = loss_api(prediction, labels)
        total_loss += loss.item() * labels.shape[0]
        loss.backward()
        optimizer.step()

        if (batch_index + 1) % 300 == 0 or (
            batch_index + 1 == len(loaded_data)
        ):
            print(
                f"Epoch {epoch + 1}/{epochs}, "
                f"Batch {batch_index + 1}/{len(loaded_data)}, "
                f"batch_loss={loss.item():.4f}"
            )
    acc = correct_num / total_num
    av_loss = total_loss / total_num
    print(
            f"Epoch {epoch + 1}: "
            f"correct={correct_num}, total={total_num}, "
            f"accuracy={acc:.2%},"
            f"av_loss={av_loss:.6f}"
            )
```
