```python
import torch
import torch.nn as nn
from torchvision import datasets,transforms
from torch.utils.data import DataLoader
```

```python
train_dataset = datasets.FashionMNIST(
    root="../../data",
    train=True,
    download=True,
    transform=transforms.ToTensor(),
)

loaded_data = DataLoader(train_dataset,batch_size=32,shuffle= True,num_workers= 0 )
images , labels= next(iter(loaded_data))
flat = images.flatten(start_dim = 1)
print(flat.shape)
```

```python
class imageMLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(784,128)
        self.layer2 = nn.Linear(128,10)

    def forward(self, x):
        x = x.flatten(start_dim = 1)
        y1 = self.layer1(x)
        y2 = torch.relu(y1)
        y3 = self.layer2(y2)
        return y3

model = imageMLP()
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
loss_api = nn.BCEWithLogitsLoss()

for step in range(5000):
    optimizer.zero_grad()
    prediction = model(flat)
    loss = loss_api(prediction,labels)
    loss.backward()
    optimizer.step()
    if step == 4999:        #我只看最后一批的预测结果
        with torch.no_grad():
            prediction = model(images)
            loss = loss_api(prediction, labels)
            print("Loss", loss)
            print("Prediction",prediction)
            print("True_labels",labels)

            
        
```
